```python
# Water Jug Problem using DFS
def water_jug_dfs(jug1_capacity, jug2_capacity, target):
    # Stack for DFS: each element is (jug1, jug2, path)
    stack = [(0, 0, [])]
    visited = set()
    
    while stack:
        jug1, jug2, path = stack.pop()
        if (jug1, jug2) in visited:
            continue
        visited.add((jug1, jug2))
        
        # Check if goal reached
        if jug1 == target or jug2 == target:
            print(f"Goal reached: Jug1 = {jug1}, Jug2 = {jug2}")
            for step in path:
                print(step)
            return
        
        # Possible moves
        moves = []

        # 1. Fill Jug1
        if jug1 < jug1_capacity:
            moves.append((jug1_capacity, jug2, "Fill Jug1"))

        # 2. Fill Jug2
        if jug2 < jug2_capacity:
            moves.append((jug1, jug2_capacity, "Fill Jug2"))

        # 3. Empty Jug1
        if jug1 > 0:
            moves.append((0, jug2, "Empty Jug1"))

        # 4. Empty Jug2
        if jug2 > 0:
            moves.append((jug1, 0, "Empty Jug2"))

        # 5. Pour Jug1 -> Jug2
        if jug1 > 0 and jug2 < jug2_capacity:
            pour_amount = min(jug1, jug2_capacity - jug2)
            moves.append((jug1 - pour_amount, jug2 + pour_amount, f"Pour Jug1 -> Jug2"))

        # 6. Pour Jug2 -> Jug1
        if jug2 > 0 and jug1 < jug1_capacity:
            pour_amount = min(jug2, jug1_capacity - jug1)
            moves.append((jug1 + pour_amount, jug2 - pour_amount, f"Pour Jug2 -> Jug1"))

        # Add all moves to stack
        for new_jug1, new_jug2, action in moves:
            stack.append((new_jug1, new_jug2, path + [f"{action}: ({new_jug1}, {new_jug2})"]))

# Example usage
water_jug_dfs(4, 3, 2)
```
